```python
import pandas as pd
from huggingface_hub import hf_hub_download

# Download the content dimension
content_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="dim_content.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

content_df = pd.read_parquet(content_path)

print("Shape:", content_df.shape)
print("\nColumns:")
print(content_df.columns.tolist())

print("\nFirst 5 rows:")
display(content_df.head())
```

```python
query_path = hf_hub_download(
    repo_id="FlyRank/internship-warehouse",
    filename="fact_content_query_90d.parquet",
    repo_type="dataset",
    token=HF_TOKEN
)

query_df = pd.read_parquet(query_path)

print("Shape:", query_df.shape)
print("\nColumns:")
print(query_df.columns.tolist())

print("\nFirst 5 rows:")
display(query_df.head())
```

```python
# Aggregate query-level performance to one row per content page

performance_df = (
    query_df
    .groupby(["client_hash_id", "content_hash_id"], as_index=False)
    .agg(
        impressions_90d=("impressions_90d", "sum"),
        clicks_90d=("clicks_90d", "sum"),
        impressions_last30=("impressions_last30", "sum"),
        clicks_last30=("clicks_last30", "sum"),
        impressions_prev30=("impressions_prev30", "sum"),
        clicks_prev30=("clicks_prev30", "sum"),
        avg_position_90d=("avg_position_90d", "mean"),
        avg_position_last30=("avg_position_last30", "mean"),
        avg_position_prev30=("avg_position_prev30", "mean"),
        content_total_impressions_90d=("content_total_impressions_90d", "first"),
        content_visible_query_count=("content_visible_query_count", "first"),
    )
)

# Calculate CTR from aggregated clicks and impressions
performance_df["ctr_90d"] = (
    performance_df["clicks_90d"] /
    performance_df["impressions_90d"].replace(0, pd.NA)
)

performance_df["ctr_last30"] = (
    performance_df["clicks_last30"] /
    performance_df["impressions_last30"].replace(0, pd.NA)
)

performance_df["ctr_prev30"] = (
    performance_df["clicks_prev30"] /
    performance_df["impressions_prev30"].replace(0, pd.NA)
)

print("Shape:", performance_df.shape)
display(performance_df.head())

```

```python
# Merge content metadata with page-level performance

df = content_df.merge(
    performance_df,
    on=["client_hash_id", "content_hash_id"],
    how="inner"
)

print("Final dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nSample:")
display(df.head())

```

```python
# Signal 1: Staleness
# Calculate how many days have passed since the content was last updated.

OBSERVATION_DATE = pd.Timestamp("2026-06-30")

df["content_updated_date"] = pd.to_datetime(
    df["content_updated_date"],
    errors="coerce"
)

df["days_since_update"] = (
    OBSERVATION_DATE - df["content_updated_date"]
).dt.days

print("Missing days_since_update:",
      df["days_since_update"].isna().sum())

print("\nStaleness summary:")
print(df["days_since_update"].describe())

```

```python
# Signal Check 1: Staleness buckets

df["staleness_bucket"] = pd.cut(
    df["days_since_update"],
    bins=[-1, 30, 60, 90, 180, float("inf")],
    labels=["0-30", "31-60", "61-90", "91-180", "180+"]
)

staleness_check = (
    df.groupby("staleness_bucket", observed=False)
      .agg(
          n=("content_hash_id", "count"),
          avg_impressions_90d=("impressions_90d", "mean"),
          avg_ctr_90d=("ctr_90d", "mean"),
          avg_position_90d=("avg_position_90d", "mean")
      )
      .reset_index()
)

# Round for readability
staleness_check["avg_impressions_90d"] = (
    staleness_check["avg_impressions_90d"].round(2)
)

staleness_check["avg_ctr_90d"] = (
    staleness_check["avg_ctr_90d"].round(4)
)

staleness_check["avg_position_90d"] = (
    staleness_check["avg_position_90d"].round(2)
)

print("Staleness signal check:")
display(staleness_check)
```

**Signal 1: Staleness — MIXED**

The staleness signal shows a mixed relationship with observed content performance. Pages updated 61–90 days ago have the weakest average CTR (0.06%) and the lowest average position performance (54.30), but performance does not continue to worsen consistently as content becomes older. The 91–180 day group has a higher average CTR (0.12%) and substantially better average position (29.51), while the 180+ group is also small (n=299). Therefore, staleness alone is not a sufficiently reliable monotonic signal for prioritizing refresh work and should not be used as the sole basis of the baseline score.


```python
# Signal Check 2: CTR versus average search position

df["position_bucket"] = pd.cut(
    df["avg_position_90d"],
    bins=[0, 3, 5, 10, 20, 50, float("inf")],
    labels=["1-3", "4-5", "6-10", "11-20", "21-50", "50+"]
)

position_check = (
    df.groupby("position_bucket", observed=False)
      .agg(
          n=("content_hash_id", "count"),
          avg_ctr_90d=("ctr_90d", "mean"),
          avg_impressions_90d=("impressions_90d", "mean")
      )
      .reset_index()
)

position_check["avg_ctr_90d"] = (
    position_check["avg_ctr_90d"].round(4)
)

position_check["avg_impressions_90d"] = (
    position_check["avg_impressions_90d"].round(2)
)

print("CTR vs position signal check:")
display(position_check)

```

**Signal 2: CTR vs Average Position — CONFIRMED**

The CTR-versus-position check shows a clear relationship between search position and click-through rate. CTR is highest around positions 4–5 (0.40%) and generally decreases as pages move into lower-ranking positions, reaching only 0.02% for pages ranking beyond position 50. Although the 1–3 bucket is slightly higher than the 4–5 bucket, the overall pattern supports using search position as a signal for identifying pages with weaker search performance and potential review opportunities.


## 1. My rule and its reason codes

*Write the rule in plain words first. Then the reason codes it can output.*

```python
# Inspect candidate thresholds for the baseline rule

rule_summary = df[
    ["impressions_90d", "ctr_90d", "avg_position_90d"]
].describe(percentiles=[0.25, 0.50, 0.75, 0.90, 0.95])

display(rule_summary)
```

```python

# Candidate baseline score distribution
# Thresholds are based on the 75th percentiles observed above.

IMPRESSIONS_THRESHOLD = 702
POSITION_THRESHOLD = 45.180609
CTR_THRESHOLD = 0.000882

df["baseline_score"] = (
    (df["avg_position_90d"] > POSITION_THRESHOLD).astype(int) * 2
    + (df["ctr_90d"] < CTR_THRESHOLD).astype(int)
    + (df["impressions_90d"] >= IMPRESSIONS_THRESHOLD).astype(int)
)

print("Baseline score distribution:")
display(
    df["baseline_score"]
    .value_counts()
    .sort_index()
    .rename_axis("baseline_score")
    .reset_index(name="n")
)
```

### **Baseline Rule**

The baseline prioritizes content pages using three observable, decision-time signals: average search position, CTR, and 90-day impressions. The thresholds are based on the 75th percentiles observed in the current dataset rather than arbitrary fixed values.

The score is calculated as follows:

* **+2 points** when `avg_position_90d > 45.18`, indicating relatively weak search position.
* **+1 point** when `ctr_90d < 0.000882`, indicating relatively weak CTR.
* **+1 point** when `impressions_90d >= 702`, indicating meaningful search visibility.

The maximum baseline score is **4**. A higher score indicates a stronger case for manual content review.

### **Reason Code**

A single reason code is assigned based on the strongest observed issue:

* `POOR_POSITION` — page has a relatively weak average search position.
* `LOW_CTR` — page has weak CTR relative to the observed distribution.
* `LOW_VISIBILITY` — page does not meet the meaningful-impressions threshold.

### **Action Label**

The score is converted into an action:

* **4 → `REVIEW_HIGH`**
* **3 → `REVIEW`**
* **2 → `MONITOR`**
* **0–1 → `NO_IMMEDIATE_ACTION`**

This baseline is intentionally simple and interpretable. It uses only observable decision-time signals and does not use future-window outcomes or label-derived inputs.


## 2. Build the ranked queue (writes the CSV)

*Code the score, rank everything, write work/outputs/baseline_action_score.csv.*

The baseline score is converted into a ranked action queue so that pages requiring the most attention appear first. The queue uses only signals available at the decision time and excludes content with an update date after the observation date.

Pages are ranked primarily by their baseline score, with higher-scoring pages receiving higher priority. Within the same score, pages with higher 90-day impressions are ranked first because they represent greater observed search visibility.

The queue contains the baseline score, a single reason code, an action label, and the key observable performance and content signals used to support manual review.


```python
# Assign exactly ONE reason code to each page

def assign_reason(row):
    if row["avg_position_90d"] > POSITION_THRESHOLD:
        return "POOR_POSITION"
    elif row["ctr_90d"] < CTR_THRESHOLD:
        return "LOW_CTR"
    elif row["impressions_90d"] < IMPRESSIONS_THRESHOLD:
        return "LOW_VISIBILITY"
    else:
        return "NO_CLEAR_ISSUE"


df["reason_code"] = df.apply(assign_reason, axis=1)


# Convert score into one action label

def assign_action(score):
    if score == 4:
        return "REVIEW_HIGH"
    elif score == 3:
        return "REVIEW"
    elif score == 2:
        return "MONITOR"
    else:
        return "NO_IMMEDIATE_ACTION"


df["action"] = df["baseline_score"].apply(assign_action)


print("Reason code distribution:")
display(
    df["reason_code"]
    .value_counts()
    .rename_axis("reason_code")
    .reset_index(name="n")
)

print("\nAction distribution:")
display(
    df["action"]
    .value_counts()
    .rename_axis("action")
    .reset_index(name="n")
)

```

### Decision-Time Eligibility

To avoid using information that would not have been available when the baseline decision was made, only pages with `content_updated_date` on or before the observation date are included in the final queue.

The final decision-time dataset contains **108,176 eligible pages**, with the latest content update date equal to **2026-06-30**. No future-dated rows remain.


```python
# Verify that every page has exactly one score, reason code and action

print("Missing baseline scores:", df["baseline_score"].isna().sum())
print("Missing reason codes:", df["reason_code"].isna().sum())
print("Missing actions:", df["action"].isna().sum())

display(
    df[
        [
            "content_hash_id",
            "baseline_score",
            "reason_code",
            "action"
        ]
    ].head(10)
)
```

```python

# Section 2 — Build the ranked baseline queue

queue_columns = [
    "client_hash_id",
    "content_hash_id",
    "keyword_hash_id",
    "url_hash_id",
    "baseline_score",
    "reason_code",
    "action",
    "impressions_90d",
    "ctr_90d",
    "avg_position_90d",
    "impressions_last30",
    "ctr_last30",
    "avg_position_last30",
    "content_updated_date",
    "search_volume",
    "word_count"
]

baseline_queue = (
    df[queue_columns]
    .sort_values(
        by=["baseline_score", "impressions_90d"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

# Add explicit rank
baseline_queue.insert(
    0,
    "rank",
    range(1, len(baseline_queue) + 1)
)

print("Queue shape:", baseline_queue.shape)

display(baseline_queue.head(20))

```

```python
# Decision-time eligibility filter
# Only use content information available on or before the observation date.

decision_df = df[
    df["content_updated_date"] <= OBSERVATION_DATE
].copy()

print("Original rows:", len(df))
print("Decision-time eligible rows:", len(decision_df))
print("Excluded future-dated rows:", len(df) - len(decision_df))
```

```python

print("Latest content update date in decision dataset:")
print(decision_df["content_updated_date"].max())

print("\nFuture-dated rows remaining:")
print(
    (decision_df["content_updated_date"] > OBSERVATION_DATE).sum()
)

```

```python

# Rebuild baseline score using only decision-time eligible pages

decision_df["baseline_score"] = (
    (decision_df["avg_position_90d"] > POSITION_THRESHOLD).astype(int) * 2
    + (decision_df["ctr_90d"] < CTR_THRESHOLD).astype(int)
    + (decision_df["impressions_90d"] >= IMPRESSIONS_THRESHOLD).astype(int)
)


# Assign exactly ONE reason code

def assign_reason(row):
    if row["avg_position_90d"] > POSITION_THRESHOLD:
        return "POOR_POSITION"
    elif row["ctr_90d"] < CTR_THRESHOLD:
        return "LOW_CTR"
    elif row["impressions_90d"] < IMPRESSIONS_THRESHOLD:
        return "LOW_VISIBILITY"
    else:
        return "NO_CLEAR_ISSUE"


decision_df["reason_code"] = decision_df.apply(assign_reason, axis=1)


# Assign action

def assign_action(score):
    if score == 4:
        return "REVIEW_HIGH"
    elif score == 3:
        return "REVIEW"
    elif score == 2:
        return "MONITOR"
    else:
        return "NO_IMMEDIATE_ACTION"


decision_df["action"] = decision_df["baseline_score"].apply(assign_action)


print("Decision-time score distribution:")
display(
    decision_df["baseline_score"]
    .value_counts()
    .sort_index()
    .rename_axis("baseline_score")
    .reset_index(name="n")
)
```

### Ranking Logic

The final queue is sorted using two criteria:

1. **Baseline score — descending:** higher scores indicate stronger evidence for review.
2. **90-day impressions — descending:** when pages have the same score, pages with greater search visibility are prioritized.

A rank number is then assigned to every eligible page, producing a reproducible review queue.


```python

# Build the final decision-time ranked queue

baseline_queue = (
    decision_df[queue_columns]
    .sort_values(
        by=["baseline_score", "impressions_90d"],
        ascending=[False, False]
    )
    .reset_index(drop=True)
)

baseline_queue.insert(
    0,
    "rank",
    range(1, len(baseline_queue) + 1)
)

print("Final decision-time queue shape:", baseline_queue.shape)

display(baseline_queue.head(10))

```

```python

# Save the final decision-time queue

baseline_queue.to_csv(
    output_path,
    index=False
)

print(f"✅ Final baseline queue written to: {output_path}")
print(f"Rows written: {len(baseline_queue):,}")

```
